```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer, load_diabetes
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import accuracy_score, classification_report, mean_squared_error, r2_score
from scipy.stats import randint, uniform
import time
from sklearn.pipeline import Pipeline
```

```python
# Set random seed for reproducibility
np.random.seed(42)
```

```python
# Load a classification dataset
cancer = load_breast_cancer()
X_class = cancer.data
y_class = cancer.target
```

```python
# Load a regression dataset
diabetes = load_diabetes()
X_reg = diabetes.data
y_reg = diabetes.target
```

```python
# Split data into training and testing sets
X_class_train, X_class_test, y_class_train, y_class_test = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42)

X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42)
```

```python
# Standardize features
scaler = StandardScaler()
X_class_train_scaled = scaler.fit_transform(X_class_train)
X_class_test_scaled = scaler.transform(X_class_test)
```

```python
X_reg_train_scaled = scaler.fit_transform(X_reg_train)
X_reg_test_scaled = scaler.transform(X_reg_test)
```

# Part 1: Grid Search for SVM Classifier

```python
# Define the parameter grid
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [0.001, 0.01, 0.1, 1],
    'kernel': ['rbf', 'poly']
}
```

```python
# Create SVM classifier
svm = SVC(random_state=42)
```

```python
# Create GridSearchCV object
grid_search = GridSearchCV(
    estimator=svm,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    verbose=0,
    scoring='accuracy',
    return_train_score=True
)
```

```python
# Start timing
start_time = time.time()
```

```python
# Fit GridSearchCV
grid_search.fit(X_class_train_scaled, y_class_train)
```

```python
# End timing
grid_time = time.time() - start_time
```

```python
# Print results
print(f"Grid Search completed in {grid_time:.2f} seconds")
print(f"Best parameters: {grid_search.best_params_}")
print(f"Best cross-validation score: {grid_search.best_score_:.4f}")
```

```python
# Evaluate on test set
y_pred = grid_search.predict(X_class_test_scaled)
grid_accuracy = accuracy_score(y_class_test, y_pred)
print(f"Test accuracy with optimal parameters: {grid_accuracy:.4f}")
```

```python
# Visualization of Grid Search results
grid_results = pd.DataFrame(grid_search.cv_results_)
grid_pivot = grid_results.pivot_table(
    values='mean_test_score',
    index='param_C',
    columns='param_gamma'
)
```

# Part 5: Learning Curves for different hyperparameter settings

```python
def plot_learning_curve(estimator, X, y, title, cv=5, train_sizes=np.linspace(0.1, 1.0, 5)):
    train_sizes, train_scores, test_scores = learning_curve(
        estimator, X, y, cv=cv, n_jobs=-1, train_sizes=train_sizes, scoring='accuracy')

    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)

    plt.figure(figsize=(10, 6))
    plt.title(title)
    plt.xlabel("Training examples")
    plt.ylabel("Score")
    plt.grid()

    plt.fill_between(train_sizes, train_scores_mean - train_scores_std,
                     train_scores_mean + train_scores_std, alpha=0.1, color="r")
    plt.fill_between(train_sizes, test_scores_mean - test_scores_std,
                     test_scores_mean + test_scores_std, alpha=0.1, color="g")
    plt.plot(train_sizes, train_scores_mean, 'o-', color="r", label="Training score")
    plt.plot(train_sizes, test_scores_mean, 'o-', color="g", label="Cross-validation score")

    plt.legend(loc="best")
    return plt
```

# Part 6: Cross-validation visualization for different hyperparameters

```python
# Function to run cross-validation with different hyperparameters and visualize results
def plot_cv_results(param_name, param_range, estimator, X, y, cv=5, scoring='accuracy'):
    cv_scores = []
    for param_value in param_range:
        params = {param_name: param_value}
        estimator.set_params(**params)
        scores = cross_val_score(estimator, X, y, cv=cv, scoring=scoring)
        cv_scores.append((param_value, np.mean(scores), np.std(scores)))

    cv_df = pd.DataFrame(cv_scores, columns=[param_name, 'Mean Score', 'Std Dev'])

    plt.figure(figsize=(10, 6))
    plt.errorbar(cv_df[param_name], cv_df['Mean Score'], yerr=cv_df['Std Dev'], marker='o')
    plt.title(f'Cross-validation scores for different {param_name} values')
    plt.xlabel(param_name)
    plt.ylabel('Mean CV Score')
    plt.grid(True)
    if param_name in ['C', 'gamma']:
        plt.xscale('log')
    return cv_df
```

```python
# Run CV visualization for C parameter
svm_model = SVC(kernel='rbf', gamma=0.01, random_state=42)
C_range = [0.01, 0.1, 1, 10, 100, 1000]
cv_results_C = plot_cv_results('C', C_range, svm_model, X_class_train_scaled, y_class_train)
plt.show()
```

```python
# Run CV visualization for gamma parameter
svm_model = SVC(kernel='rbf', C=1, random_state=42)
gamma_range = [0.0001, 0.001, 0.01, 0.1, 1, 10]
cv_results_gamma = plot_cv_results('gamma', gamma_range, svm_model, X_class_train_scaled, y_class_train)
plt.show()
```

```python
# ==============================================================================
# PART 1: SETUP AND DATA PREPARATION
# ==============================================================================
# Import the necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import time

# Load the California housing dataset
housing = fetch_california_housing()
X = housing.data
y = housing.target
feature_names = housing.feature_names

# Split the dataset into training and testing sets (80/20 split)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Training samples: {X_train.shape[0]}")
print(f"Testing samples: {X_test.shape[0]}")
print(f"Number of features: {X_train.shape[1]}")
```

```python
# ==============================================================================
# PART 2: BASELINE MODELS
# ==============================================================================
from sklearn.linear_model import Ridge
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# Create baseline models with default parameters
baseline_models = {
    "Ridge": Ridge(),
    "SVR": SVR(),
    "Random Forest": RandomForestRegressor(random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42)
}

# Use scaled features for all models here.
# Tree-based models are not sensitive to feature scaling, so scaling does not
# materially change their behavior.
baseline_results = []

for name, model in baseline_models.items():
    start_time = time.time()

    rmse_scores = -cross_val_score(
        model,
        X_train_scaled,
        y_train,
        cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )

    r2_scores = cross_val_score(
        model,
        X_train_scaled,
        y_train,
        cv=5,
        scoring="r2",
        n_jobs=-1
    )

    elapsed = time.time() - start_time

    baseline_results.append({
        "Model": name,
        "Mean RMSE": rmse_scores.mean(),
        "Std RMSE": rmse_scores.std(),
        "Mean R²": r2_scores.mean(),
        "Std R²": r2_scores.std(),
        "Time (s)": elapsed
    })

baseline_results_df = pd.DataFrame(baseline_results)
display(baseline_results_df)

# Visualize baseline model performance
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=baseline_results_df, x="Model", y="Mean RMSE", ax=axes[0])
axes[0].set_title("Baseline RMSE")
axes[0].tick_params(axis="x", rotation=25)

sns.barplot(data=baseline_results_df, x="Model", y="Mean R²", ax=axes[1])
axes[1].set_title("Baseline R²")
axes[1].tick_params(axis="x", rotation=25)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# PART 3: HYPERPARAMETER TUNING USING GRID SEARCH
# ==============================================================================
from sklearn.model_selection import GridSearchCV

# Define parameter grids for each model
grid_params = {
    "Ridge": {
        "alpha": [0.01, 0.1, 1, 10, 100]
    },
    "SVR": {
        "C": [0.1, 1, 10, 100],
        "gamma": ["scale", 0.001, 0.01, 0.1],
        "kernel": ["rbf", "poly"]
    },
    "Random Forest": {
        "n_estimators": [100, 200],
        "max_depth": [None, 10, 20],
        "min_samples_split": [2, 5, 10]
    },
    "Gradient Boosting": {
        "n_estimators": [100, 200],
        "max_depth": [2, 3, 5],
        "learning_rate": [0.01, 0.05, 0.1]
    }
}

def perform_grid_search(model, param_grid, X_data=X_train_scaled, y_data=y_train):
    start_time = time.time()

    search = GridSearchCV(
        estimator=model,
        param_grid=param_grid,
        cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1,
        refit=True
    )

    search.fit(X_data, y_data)
    elapsed = time.time() - start_time

    return (
        search.best_estimator_,
        search.best_params_,
        -search.best_score_,
        elapsed
    )

grid_results = {}
grid_rows = []

for name, model in baseline_models.items():
    best_model, best_params, best_rmse, elapsed = perform_grid_search(
        model,
        grid_params[name]
    )

    grid_results[name] = {
        "model": best_model,
        "params": best_params,
        "rmse": best_rmse,
        "time": elapsed
    }

    grid_rows.append({
        "Model": name,
        "Best RMSE": best_rmse,
        "Time (s)": elapsed,
        "Best Parameters": best_params
    })

grid_results_df = pd.DataFrame(grid_rows)
display(grid_results_df)

plt.figure(figsize=(10, 5))
sns.barplot(data=grid_results_df, x="Model", y="Best RMSE")
plt.title("Grid Search: Best Cross-Validation RMSE")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# PART 4: HYPERPARAMETER TUNING USING RANDOM SEARCH
# ==============================================================================
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform, randint

# Define parameter distributions
random_params = {
    "Ridge": {
        "alpha": uniform(0.01, 100)
    },
    "SVR": {
        "C": uniform(0.1, 100),
        "gamma": uniform(0.001, 0.2),
        "kernel": ["rbf", "poly"]
    },
    "Random Forest": {
        "n_estimators": randint(100, 301),
        "max_depth": [None, 10, 20, 30],
        "min_samples_split": randint(2, 11)
    },
    "Gradient Boosting": {
        "n_estimators": randint(100, 301),
        "max_depth": randint(2, 7),
        "learning_rate": uniform(0.01, 0.19)
    }
}

def perform_random_search(model, param_distributions, n_iter=20,
                          X_data=X_train_scaled, y_data=y_train):
    start_time = time.time()

    search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_distributions,
        n_iter=n_iter,
        cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1,
        random_state=42,
        refit=True
    )

    search.fit(X_data, y_data)
    elapsed = time.time() - start_time

    return (
        search.best_estimator_,
        search.best_params_,
        -search.best_score_,
        elapsed
    )

random_results = {}
random_rows = []

for name, model in baseline_models.items():
    best_model, best_params, best_rmse, elapsed = perform_random_search(
        model,
        random_params[name],
        n_iter=20
    )

    random_results[name] = {
        "model": best_model,
        "params": best_params,
        "rmse": best_rmse,
        "time": elapsed
    }

    random_rows.append({
        "Model": name,
        "Best RMSE": best_rmse,
        "Time (s)": elapsed,
        "Best Parameters": best_params
    })

random_results_df = pd.DataFrame(random_rows)
display(random_results_df)

plt.figure(figsize=(10, 5))
sns.barplot(data=random_results_df, x="Model", y="Best RMSE")
plt.title("Random Search: Best Cross-Validation RMSE")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# PART 5: HYPERPARAMETER TUNING USING BAYESIAN OPTIMIZATION
# ==============================================================================
# Install scikit-optimize if not already installed
# Run this once in a notebook if needed:
# !pip install scikit-optimize

from skopt import BayesSearchCV
from skopt.space import Real, Integer, Categorical

# Define search spaces for each model
bayes_spaces = {
    "Ridge": {
        "alpha": Real(0.01, 100, prior="log-uniform")
    },
    "SVR": {
        "C": Real(0.1, 100, prior="log-uniform"),
        "gamma": Real(0.001, 1.0, prior="log-uniform"),
        "kernel": Categorical(["rbf", "poly"])
    },
    "Random Forest": {
        "n_estimators": Integer(100, 300),
        "max_depth": Integer(5, 30),
        "min_samples_split": Integer(2, 10)
    },
    "Gradient Boosting": {
        "n_estimators": Integer(100, 300),
        "max_depth": Integer(2, 6),
        "learning_rate": Real(0.01, 0.2, prior="log-uniform")
    }
}

def perform_bayesian_search(model, search_space, n_iter=20,
                             X_data=X_train_scaled, y_data=y_train):
    start_time = time.time()

    search = BayesSearchCV(
        estimator=model,
        search_spaces=search_space,
        n_iter=n_iter,
        cv=5,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1,
        random_state=42,
        refit=True
    )

    search.fit(X_data, y_data)
    elapsed = time.time() - start_time

    return (
        search.best_estimator_,
        search.best_params_,
        -search.best_score_,
        elapsed
    )

bayes_results = {}
bayes_rows = []

for name, model in baseline_models.items():
    best_model, best_params, best_rmse, elapsed = perform_bayesian_search(
        model,
        bayes_spaces[name],
        n_iter=20
    )

    bayes_results[name] = {
        "model": best_model,
        "params": best_params,
        "rmse": best_rmse,
        "time": elapsed
    }

    bayes_rows.append({
        "Model": name,
        "Best RMSE": best_rmse,
        "Time (s)": elapsed,
        "Best Parameters": best_params
    })

bayes_results_df = pd.DataFrame(bayes_rows)
display(bayes_results_df)

plt.figure(figsize=(10, 5))
sns.barplot(data=bayes_results_df, x="Model", y="Best RMSE")
plt.title("Bayesian Optimization: Best Cross-Validation RMSE")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# PART 6: COMPARISON OF TUNING METHODS
# ==============================================================================

# Create a comprehensive comparison table
comparison_rows = []

for model_name in baseline_models:
    baseline_row = baseline_results_df[
        baseline_results_df["Model"] == model_name
    ].iloc[0]

    for method, results_dict in [
        ("Grid Search", grid_results),
        ("Random Search", random_results),
        ("Bayesian Optimization", bayes_results)
    ]:
        result = results_dict[model_name]
        improvement = baseline_row["Mean RMSE"] - result["rmse"]

        comparison_rows.append({
            "Model": model_name,
            "Method": method,
            "Best RMSE": result["rmse"],
            "Time (s)": result["time"],
            "Improvement over Baseline": improvement,
            "Improvement (%)": (
                improvement / baseline_row["Mean RMSE"] * 100
            )
        })

comparison_df = pd.DataFrame(comparison_rows)
display(comparison_df)

# Visualize RMSE comparison
plt.figure(figsize=(12, 6))
sns.barplot(
    data=comparison_df,
    x="Model",
    y="Best RMSE",
    hue="Method"
)
plt.title("Comparison of Hyperparameter Tuning Methods")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

# Visualize tuning time
plt.figure(figsize=(12, 6))
sns.barplot(
    data=comparison_df,
    x="Model",
    y="Time (s)",
    hue="Method"
)
plt.title("Tuning Time Comparison")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# PART 7: ANALYZE TUNING EFFICIENCY AND META-LEARNING
# ==============================================================================

# Calculate efficiency:
# efficiency = improvement in RMSE / tuning time
# Higher values indicate more improvement per second.
comparison_df["Efficiency"] = (
    comparison_df["Improvement over Baseline"] /
    comparison_df["Time (s)"].replace(0, np.nan)
)

efficiency_summary = (
    comparison_df
    .groupby("Method")["Efficiency"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

display(efficiency_summary)

plt.figure(figsize=(9, 5))
sns.barplot(data=efficiency_summary, x="Method", y="Efficiency")
plt.title("Average Tuning Efficiency")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Analyze relationship between number of hyperparameters and efficiency
n_hyperparameters = {
    "Grid Search": {
        "Ridge": 1, "SVR": 3, "Random Forest": 3, "Gradient Boosting": 3
    },
    "Random Search": {
        "Ridge": 1, "SVR": 3, "Random Forest": 3, "Gradient Boosting": 3
    },
    "Bayesian Optimization": {
        "Ridge": 1, "SVR": 3, "Random Forest": 3, "Gradient Boosting": 3
    }
}

comparison_df["Number of Hyperparameters"] = [
    n_hyperparameters[row["Method"]][row["Model"]]
    for _, row in comparison_df.iterrows()
]

plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=comparison_df,
    x="Number of Hyperparameters",
    y="Efficiency",
    hue="Method",
    style="Model",
    s=100
)
plt.title("Hyperparameter Count vs Tuning Efficiency")
plt.tight_layout()
plt.show()

# --------------------------------------------------------------------------
# Meta-learning: extract dataset characteristics
# --------------------------------------------------------------------------
def extract_dataset_characteristics(X_data, y_data):
    X_data = np.asarray(X_data)
    y_data = np.asarray(y_data)

    characteristics = {
        "n_samples": X_data.shape[0],
        "n_features": X_data.shape[1],
        "mean_feature_mean": np.mean(np.mean(X_data, axis=0)),
        "mean_feature_std": np.mean(np.std(X_data, axis=0)),
        "mean_feature_min": np.mean(np.min(X_data, axis=0)),
        "mean_feature_max": np.mean(np.max(X_data, axis=0)),
        "target_mean": np.mean(y_data),
        "target_std": np.std(y_data),
        "target_min": np.min(y_data),
        "target_max": np.max(y_data)
    }

    return characteristics

# Generate small datasets and determine which algorithm performs best
rng = np.random.RandomState(42)
meta_rows = []

meta_models = {
    "Ridge": Ridge(),
    "SVR": SVR(),
    "Random Forest": RandomForestRegressor(
        n_estimators=100, random_state=42, n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42)
}

for sample_id in range(12):
    sample_size = rng.randint(1000, 4001)
    indices = rng.choice(len(X_train_scaled), size=sample_size, replace=False)

    X_sample = X_train_scaled[indices]
    y_sample = y_train[indices]

    characteristics = extract_dataset_characteristics(X_sample, y_sample)

    model_scores = {}
    for model_name, model in meta_models.items():
        score = -cross_val_score(
            model,
            X_sample,
            y_sample,
            cv=3,
            scoring="neg_root_mean_squared_error",
            n_jobs=-1
        ).mean()

        model_scores[model_name] = score

    best_algorithm = min(model_scores, key=model_scores.get)

    characteristics["best_algorithm"] = best_algorithm
    characteristics["sample_id"] = sample_id
    meta_rows.append(characteristics)

meta_dataset = pd.DataFrame(meta_rows)
display(meta_dataset)

# Train a classifier to predict the best algorithm
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

meta_features = [
    "n_samples", "n_features", "mean_feature_mean",
    "mean_feature_std", "mean_feature_min", "mean_feature_max",
    "target_mean", "target_std", "target_min", "target_max"
]

X_meta = meta_dataset[meta_features]
y_meta = meta_dataset["best_algorithm"]

if len(meta_dataset) >= 8 and y_meta.nunique() >= 2:
    X_meta_train, X_meta_test, y_meta_train, y_meta_test = train_test_split(
        X_meta,
        y_meta,
        test_size=0.30,
        random_state=42,
        stratify=y_meta
    )

    meta_classifier = RandomForestClassifier(
        n_estimators=100,
        random_state=42
    )
    meta_classifier.fit(X_meta_train, y_meta_train)

    meta_predictions = meta_classifier.predict(X_meta_test)

    print("Meta-learning accuracy:",
          accuracy_score(y_meta_test, meta_predictions))
    print(classification_report(
        y_meta_test,
        meta_predictions,
        zero_division=0
    ))
else:
    print(
        "Not enough variation in the sampled datasets/classes to train "
        "a meaningful meta-classifier."
    )
```

```python
# ==============================================================================
# PART 8: ENSEMBLE OF TUNED MODELS
# ==============================================================================

# Create an ensemble using the best tuned model from each tuning method.
# We select the best model for each algorithm based on CV RMSE.
all_tuned_results = {
    "Grid Search": grid_results,
    "Random Search": random_results,
    "Bayesian Optimization": bayes_results
}

best_models = {}

for model_name in baseline_models:
    candidates = []

    for method_name, result_dict in all_tuned_results.items():
        result = result_dict[model_name]
        candidates.append((result["rmse"], method_name, result["model"]))

    best_rmse, best_method, best_model = min(candidates, key=lambda x: x[0])

    best_models[model_name] = best_model

    print(
        f"{model_name}: selected {best_method} "
        f"(CV RMSE = {best_rmse:.4f})"
    )

# Fit the selected tuned models on the full training data
for model in best_models.values():
    model.fit(X_train_scaled, y_train)

# Evaluate individual models on the untouched test set
individual_test_results = []

test_predictions = {}

for model_name, model in best_models.items():
    predictions = model.predict(X_test_scaled)
    test_predictions[model_name] = predictions

    individual_test_results.append({
        "Model": model_name,
        "Test RMSE": mean_squared_error(
            y_test, predictions, squared=False
        ),
        "Test R²": r2_score(y_test, predictions)
    })

individual_test_df = pd.DataFrame(individual_test_results)
display(individual_test_df)

# Simple averaging ensemble
ensemble_prediction = np.mean(
    np.column_stack(list(test_predictions.values())),
    axis=1
)

ensemble_rmse = mean_squared_error(
    y_test, ensemble_prediction, squared=False
)
ensemble_r2 = r2_score(y_test, ensemble_prediction)

print(f"Ensemble Test RMSE: {ensemble_rmse:.4f}")
print(f"Ensemble Test R²: {ensemble_r2:.4f}")

# Compare individual models with ensemble
ensemble_comparison = pd.concat([
    individual_test_df,
    pd.DataFrame([{
        "Model": "Ensemble",
        "Test RMSE": ensemble_rmse,
        "Test R²": ensemble_r2
    }])
], ignore_index=True)

display(ensemble_comparison)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=ensemble_comparison,
    x="Model",
    y="Test RMSE"
)
plt.title("Individual Tuned Models vs Ensemble")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# PART 10: CONCLUSION AND REPORTING
# ==============================================================================

# Identify the best individual model and tuning method
best_individual = individual_test_df.loc[
    individual_test_df["Test RMSE"].idxmin()
]

best_tuning_row = comparison_df.loc[
    comparison_df["Best RMSE"].idxmin()
]

most_efficient_method = efficiency_summary.iloc[0]

ensemble_is_better = ensemble_rmse < best_individual["Test RMSE"]

print("=" * 80)
print("FINAL CONCLUSION")
print("=" * 80)

print(
    f"Best individual model on the test set: "
    f"{best_individual['Model']}"
)
print(
    f"Best individual test RMSE: "
    f"{best_individual['Test RMSE']:.4f}"
)
print(
    f"Best individual test R²: "
    f"{best_individual['Test R²']:.4f}"
)

print(
    f"\nBest tuning result according to cross-validation: "
    f"{best_tuning_row['Model']} using {best_tuning_row['Method']}"
)
print(
    f"Cross-validation RMSE: "
    f"{best_tuning_row['Best RMSE']:.4f}"
)

print(
    f"\nMost efficient tuning method on average: "
    f"{most_efficient_method['Method']}"
)
print(
    f"Average efficiency: "
    f"{most_efficient_method['Efficiency']:.6f}"
)

print(
    f"\nEnsemble RMSE: {ensemble_rmse:.4f}"
)
print(
    f"Ensemble R²: {ensemble_r2:.4f}"
)
print(
    f"Did the ensemble outperform the best individual model? "
    f"{'Yes' if ensemble_is_better else 'No'}"
)

print("\nKey lessons:")
print("1. Grid Search is exhaustive but can become expensive as the search space grows.")
print("2. Random Search can explore large spaces efficiently by sampling configurations.")
print("3. Bayesian Optimization uses previous evaluations to focus on promising regions.")
print("4. Log-uniform spaces are useful for parameters such as C and gamma.")
print("5. Ensemble predictions can benefit from combining models with different errors.")
print("6. Meta-learning requires many diverse datasets to produce reliable predictions.")
```
